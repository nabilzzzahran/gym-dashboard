```python
PR = {}
workout_data = [
    {"exercise": "Bench Press", "weight": 125, "month": "September"},
    {"exercise": "Bench Press", "weight": 130, "month": "October"},
    {"exercise": "Squat", "weight": 185, "month": "September"},
]
for workout in workout_data:
    exercise = workout["exercise"]
    weight = workout["weight"]
    month = workout["month"]
    if exercise not in PR or weight > PR[exercise]["weight"]:
        PR[exercise] = {"weight": weight, "month": month}


for exercise, data in PR.items():
    print( f"PR for {exercise}: {data['weight']} at {data['month']}")
```
